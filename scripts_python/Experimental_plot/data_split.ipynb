```python
# 读取电压、容量曲线，注意要把标题去掉
curve_data = pd.read_csv('experiment-data/T2-pristine-0.1c-4-4.65.csv')
# curve_data.columns
# df_test = curve_data[curve_data['Cycle Index'] == 3]
# test_data = df_test[df_test['Step Type'] == "CC Chg"][["Voltage", "Spec. Cap."]]
# test_columns = ["Voltage_charge_1", "Capacity_charge_1",
#                    "Voltage_discharge_1", "Capacity_discharge_1"]
# test_data.rename(columns={'Voltage': test_columns[0], "Spec. Cap.": test_columns[1]}, inplace=True)
curve_data
# curve_data[["Voltage", "Spec. Cap."]]
# curve_data.iloc[:, 3:5]
```

```python
cycle_number = len(set(curve_data["Cycle Index"].values))
# 根据cycle_index判断有多少个循环，用于后续分布列
cycle_number
```

```python
#
export_cycle = [1
    # , 6, 11, 16, 21, 26, 31, 36, 41
                ]
# 选取的循环，分别是0.5c 2c， 5c， 10c， 20c， 40c， 60c， 100c， 0.5c
cycle_data = pd.DataFrame(data=None)
for i in export_cycle: # 循环数设置,这一行是所有数据都选取

    new_columns = ["Voltage_charge_"+str(i), "Capacity_charge_"+str(i),
                   "Voltage_discharge_"+str(i), "Capacity_discharge_"+str(i)] # 每个循环的数据设计成
    df_cycle = curve_data[curve_data['Cycle Index'] == i]  # 选取第i圈循环的数据
    df_charge_i = df_cycle[df_cycle['Step Type'] == "CC Chg"][["Voltage", "Spec. Cap."]].reset_index(drop=True)
    df_charge_i.rename(columns={'Voltage': new_columns[0], "Spec. Cap.": new_columns[1]}, inplace=True)
    # print(df_charge_i)
    df_discharge_i = df_cycle[df_cycle['Step Type'] == "CC DChg"][["Voltage", "Spec. Cap."]].reset_index(drop=True)
    df_discharge_i.rename(columns={'Voltage': new_columns[2], "Spec. Cap.": new_columns[3]}, inplace=True)
    cycle_data = pd.concat([cycle_data, df_charge_i, df_discharge_i], axis=1)
cycle_data
```
